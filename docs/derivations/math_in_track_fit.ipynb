```python
import sympy as sp

# Define symbolic variables
dx, dy, dt = sp.symbols("dx dy dt")  # residuals
tx, ty, tb = sp.symbols("tx ty tb")  # track slopes
Cxx, Cyy, Ctt = sp.symbols("Cxx Cyy Ctt")  # diagonal elements only
dz = sp.symbols("dz")

# Residual vector
res = sp.Matrix([dx, dy, dt])  # 3x1

# Covariance matrix (diagonal first)
Cov = sp.diag(Cxx, Cyy, Ctt)

# Compute inverse
invCov = Cov.inv()

# Construct m_H matrix (4x3)
m_H = sp.Matrix([
    [-1,  0,  0],     # d r(x) / d v(x)
    [ 0, -1,  0],     # d r(y) / d v(y)
    [tx, ty, tb],     # d r(x,y,t) / d v(z)
    [ 0,  0, -1],     # d r(t) / d v(t)
])

# Calculate similarity: res^T * invCov * res
similarity = (res.T * invCov * res)

# Calculate halfD2Chi2DX2 = H * invCov * H^T
halfD2Chi2DX2 = m_H * invCov * m_H.T

# Calculate first derivative vector: H * invCov * res
halfDChi2DX = m_H * invCov * res

halfD2Chi2DX2
```
